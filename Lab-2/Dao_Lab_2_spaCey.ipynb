```python
nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")

doc = nlp(text)

sentences = list(doc.sents)

print("Number of sentences:", len(sentences))

print("\nFirst 3 sentences:")
for sent in sentences[:3]:
    print(sent)
```

Got the total amount of words: 2911
and unique words: 738



```python
words = [token.text.lower() for token in doc if token.is_alpha]

print("Total words:", len(words))
print("Unique words:", len(set(words)))
```

We then got the most frequently used words. "The" and "To" are the most used words with 69 and 65 uses.

```python
from collections import Counter

word_freq = Counter(words)

print("Top 10 most frequent words:")
for word, count in word_freq.most_common(10):
    print(word, count)
```

Ran a full spaCy pipeline to find how many named entities there were and what kinds of entities appeared. There are 104. However, some of the entities don't make sense to me at least. For example, "REM" is counted as a person. The word "BASELINE" is also counted as a work or art or person multiple times.


```python
nlp2 = spacy.load("en_core_web_sm")

doc2 = nlp2(text)

print("Named Entities:")
total_named_entities = 0
for ent in doc2.ents:
    print(ent.text, ent.label_)
    total_named_entities += 1

print("Total Named Entities:", total_named_entities)
```

Used PhraseMatcher

My original topic of interest was looking at typography and different types of fonts. The words I chose were typography, letter strokes, hierarchy:, and composition.
Because My sentences are made just using a new line, you don't get the full context of each sentence.
The transcript only contains 2 of the four words: typography and hierarchy,

```python
from spacy.matcher import PhraseMatcher

matcher = PhraseMatcher(nlp2.vocab, attr="LOWER")

phrases = ["typography", "letter stokes", "hierarchy", "composition"]

patterns = [nlp2(p) for p in phrases]

matcher.add("TECH_TERMS", patterns)

matches = matcher(doc2)

print("Matches found:")
for match_id, start, end in matches:
    print(doc2[start:end])
    print("Sentence:", doc2[start].sent)
```
